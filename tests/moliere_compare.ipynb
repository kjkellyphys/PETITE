```python
import numpy as np
import sys
sys.path.insert(0,'../src/')

import moliere

import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# Multiple scattering in tungsten
X0 = 6.76 # g/cm^2
t_over_X0 = 0.1
beta = 0.999
A = 183.84
Z = 74.
precise_moliere_sample = [moliere.generate_moliere_angle(t_over_X0*X0, beta, A , Z, 1.) for u in range(100000)]
approx_moliere_sample = [moliere.generate_moliere_angle_simplified(t_over_X0, beta, 1.) for u in range(100000)]
```

```python
t_over_X0*X0
```

```python
bins= np.linspace(-2,2,50)

plt.hist(precise_moliere_sample,bins=bins,label=r'Full Moliere',density=True)

plt.hist(approx_moliere_sample,alpha=0.5,bins=bins,label=r'Gaussian (Highland via Lynch and Dahl)',density=True)


plt.xlabel(r'$\theta$')
#plt.xscale('log')
#plt.yscale('log')
#plt.ylim([1e-6,1.2])
plt.legend(loc='lower left')
```
